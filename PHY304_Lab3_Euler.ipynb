```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
nsteps=100000

def euler(force, times, x0, v0, m,dt):

    x = np.zeros(nsteps)
    v = np.zeros(nsteps)
    x[0]=x0
    v[0]=v0

    for i in range(nsteps-1):
        a = force(times[i], x[i], v[i]) / m
        v[i+1] = v[i] + a*dt
        x[i+1] = x[i] + v[i]*dt
    return x, v
    
times,dt = np.linspace(0, 10, nsteps,retstep=True)

# for 

x0 = 1
v0 = 0
m = 1
g = 9.8

def gravity(t,x,v):
    return -m*g

xt = x0 + v0*times - 0.5*g*times**2
vt = v0 - g*times

x,v=euler(gravity, times, x0, v0, m,dt)

#plt.plot(xt,vt)


plt.plot(times,vt)
plt.plot(times,v,ls="dashed")
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.show()

plt.plot(times,xt)
plt.plot(times,x,ls="dashed")
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.show()

k = 1

def spring(t,x,v):
    return -k*x

xs = np.cos(times) 
vs = -np.sin(times)

x,v=euler(spring,times,x0,v0,m,dt)

plt.plot(times,vs)
plt.plot(times,v,ls="dashed")
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.show()

plt.plot(times,xs)
plt.plot(times,x,ls="dashed")
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.show()
```
