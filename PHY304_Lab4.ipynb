```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

Linear: F = -bv
Fg=-mg-bv
v = -mg/b

Quadratic F = -cv|v|
Fg=-mg-cv|v|
v = (-mg/c)^1/2



```python
nsteps=100000

def euler(force, times, x0, v0, m,dt):

    x = np.zeros(nsteps)
    v = np.zeros(nsteps)
    x[0]=x0
    v[0]=v0

    for i in range(nsteps-1):
        a = force(times[i], x[i], v[i]) / m
        v[i+1] = v[i] + a*dt
        x[i+1] = x[i] + v[i]*dt
    return x, v
    
times,dt = np.linspace(0, 10, nsteps,retstep=True)

# for 

x0 = 1
v0 = 0
m = 1
g = 9.8
b = 1
c = 1

def gravity(t,x,v):
    return -m*g

def linear(t,x,v):
    return -m*g - b*v

def quad(t,x,v):
    return -m*g - c*v*abs(v)

xt = x0 + v0*times - 0.5*g*times**2
vt = v0 - g*times

xl,vl=euler(linear, times, x0, v0, m,dt)

vfl = -m*g/b

plt.plot(times,vl)
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.axhline(vfl,ls="dashed")
plt.show()

plt.plot(times,xl)
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.show()

x,v=euler(quad, times, x0, v0, m,dt)


#plt.plot(xt,vt)


vfq = -(m*g/c)**0.5

plt.plot(times,v)
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.axhline(vfq,ls="dashed")
plt.show()

plt.plot(times,x)
plt.xlabel("time (s)")
plt.ylabel("position (m)")
plt.show()
```

The velocity versus time graph with quadratic air resistance is steeper than the one with linear air resistance. It reaches terminal velocity faster.The position versus time graph with quadratic air resistance has a smaller slope.
